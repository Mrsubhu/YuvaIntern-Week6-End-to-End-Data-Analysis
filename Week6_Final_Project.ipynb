```python
from pathlib import Path
from docx import Document
from docx.shared import Inches, Pt

MAIN_DIR = Path(r"D:\Yuva Intern Tasks\Task 6")
OUTPUT_DIR = MAIN_DIR / "Output"
CHARTS_DIR = MAIN_DIR / "Charts"

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

# Create Word document
doc = Document()

# Page margins
section = doc.sections[0]
section.top_margin = Inches(0.7)
section.bottom_margin = Inches(0.7)
section.left_margin = Inches(0.8)
section.right_margin = Inches(0.8)

# Default font
doc.styles["Normal"].font.name = "Arial"
doc.styles["Normal"].font.size = Pt(10)

# Title page
doc.add_heading(
    "END-TO-END DATA ANALYSIS PROJECT",
    level=0
)
doc.add_paragraph("Online Retail Transaction Analysis")
doc.add_paragraph("YuvaIntern – Week 6 Final Project")
doc.add_paragraph("Tools: Python, Pandas, Jupyter Notebook")
doc.add_page_break()

# Section files created in previous steps
section_files = [
    ("Executive Summary", "Executive_Summary.txt"),
    ("Introduction and Objectives", "Introduction_and_Objectives.txt"),
    ("Data Exploration", "Data_Exploration_Section.txt"),
    ("Data Cleaning", "Data_Cleaning_Section.txt"),
    ("Data Visualization", "Data_Visualization_Section.txt"),
    ("Statistical Analysis", "Statistical_Analysis_Section.txt"),
    ("Machine Learning", "Machine_Learning_Section.txt"),
    ("Conclusion and Recommendations", "Conclusion_and_Recommendations.txt"),
]

for heading, filename in section_files:
    file_path = OUTPUT_DIR / filename

    doc.add_heading(heading, level=1)

    if file_path.exists():
        content = file_path.read_text(encoding="utf-8")

        for line in content.splitlines():
            line = line.strip()

            if not line:
                continue
            elif line.isupper() and len(line) < 80:
                doc.add_heading(line, level=2)
            else:
                doc.add_paragraph(line)
    else:
        doc.add_paragraph(
            f"Section file not found: {filename}"
        )

# Add available charts
chart_files = sorted(
    file for file in CHARTS_DIR.rglob("*")
    if file.is_file()
    and file.suffix.lower() in [".png", ".jpg", ".jpeg"]
)

if chart_files:
    doc.add_page_break()
    doc.add_heading("Appendix: Project Charts", level=1)

    for chart in chart_files:
        doc.add_paragraph(chart.name)
        doc.add_picture(str(chart), width=Inches(5.8))

# Save final report
report_path = OUTPUT_DIR / "YuvaIntern_Final_Data_Analysis_Report.docx"
doc.save(report_path)

print("Word report created successfully!")
print("Report location:", report_path)
print("Charts included:", len(chart_files))
print("Sections processed:", len(section_files))
```

```python
from pathlib import Path
import pandas as pd

MAIN_DIR = Path(r"D:\Yuva Intern Tasks")
OUTPUT_DIR = MAIN_DIR / "Task 6" / "Output"
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

records = []

for task_no in range(2, 6):
    task_dir = MAIN_DIR / f"Task {task_no}"

    if not task_dir.exists():
        continue

    for file in task_dir.rglob("*.csv"):
        try:
            data = pd.read_csv(file, nrows=3)

            records.append({
                "Task": f"Task {task_no}",
                "File": file.name,
                "Columns": ", ".join(data.columns.astype(str)),
                "Path": str(file)
            })
        except Exception as error:
            records.append({
                "Task": f"Task {task_no}",
                "File": file.name,
                "Columns": f"Could not read: {error}",
                "Path": str(file)
            })

inventory = pd.DataFrame(records)

if not inventory.empty:
    display(inventory)

    save_path = OUTPUT_DIR / "Task2_to_Task5_Results_Index.csv"
    inventory.to_csv(save_path, index=False)
    print("\nSaved:", save_path)
else:
    print("Task 2–5 mein CSV result files nahi mili.")
```

```python
from pathlib import Path
from docx import Document
from docx.shared import Inches

TASK6_DIR = Path(r"D:\Yuva Intern Tasks\Task 6")
CHARTS_DIR = TASK6_DIR / "Charts"
OUTPUT_DIR = TASK6_DIR / "Output"

report_path = OUTPUT_DIR / "YuvaIntern_Final_Data_Analysis_Report.docx"

if not report_path.exists():
    print("Final Word report nahi mili. Pehle report creation code run karo.")

else:
    doc = Document(report_path)

    existing_images = len(doc.inline_shapes)

    chart_files = sorted([
        file for file in CHARTS_DIR.iterdir()
        if file.is_file()
        and file.suffix.lower() in [".png", ".jpg", ".jpeg"]
    ]) if CHARTS_DIR.exists() else []

    print("Images already in report:", existing_images)
    print("Chart files available:", len(chart_files))

    if not chart_files:
        print("Task 6/Charts folder mein koi chart image nahi mili.")
        print("Task 3, Task 4 aur Task 5 ke charts copy karne honge.")

    elif existing_images > 0:
        print("Report mein images pehle se hain.")
        print("Duplicate images add nahi ki gayi.")

    else:
        doc.add_page_break()
        doc.add_heading("Appendix: Project Charts", level=1)

        for chart in chart_files:
            doc.add_paragraph(chart.name)
            doc.add_picture(str(chart), width=Inches(5.8))

        doc.save(report_path)

        print("Charts added successfully!")
        print("Updated report:", report_path)
```

```python
from pathlib import Path
from docx import Document

MAIN_DIR = Path(r"D:\Yuva Intern Tasks")
REPORT_PATH = (
    MAIN_DIR
    / "Task 6"
    / "Output"
    / "YuvaIntern_Final_Data_Analysis_Report.docx"
)

if not REPORT_PATH.exists():
    print("Final report nahi mili. Pehle report-generation cell run karo.")
else:
    doc = Document(REPORT_PATH)

    text = "\n".join(p.text for p in doc.paragraphs).lower()

    required_sections = [
        "executive summary",
        "introduction",
        "data exploration",
        "data cleaning",
        "data visualization",
        "statistical analysis",
        "machine learning",
        "conclusion",
        "recommendations"
    ]

    print("REPORT:", REPORT_PATH)
    print("File size:", round(REPORT_PATH.stat().st_size / 1024, 2), "KB")
    print("Paragraphs:", len(doc.paragraphs))
    print("Tables:", len(doc.tables))
    print("Embedded images:", len(doc.inline_shapes))
    print("\nSECTION CHECK:")

    for section in required_sections:
        found = section in text
        print(f"{'PASS' if found else 'CHECK'} - {section.title()}")

    print("\nQuality check completed.")
```
